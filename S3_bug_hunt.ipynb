## How to read an error message

Run the cell below. It will fail, on purpose.

```python
amount = int("12,500")
```

Python's error message is called a **traceback**. Read it **from the bottom up**:

1. **The last line** tells you *what* went wrong: `ValueError: invalid literal for int() with base 10: '12,500'`
   → Python can't turn the text `"12,500"` into a whole number because of the comma.
2. **The arrow `---->`** (or the `line N` above it) tells you *where* it went wrong.

If you don't understand the last line, copy it **exactly** into Google, or into an AI assistant (see the tips at the end).

## Three kinds of bug

| Kind | What you see | Example |
|---|---|---|
| **Syntax error** | Python refuses to run the code at all | a missing `:` or bracket |
| **Runtime error** | the code starts, then crashes with a traceback | dividing by zero, a misspelt variable name |
| **Logical error** | **no error at all**, just the wrong answer | `<` where you meant `>` |

Logical errors are the dangerous ones. The only way to catch them is to **check the answer**, which is what the test cells do.

## Bug 1 · Simple interest

Simple interest = principal × rate × years. So 10,000 at 5% for 1 year should be **500**.

```python
def calculate_interest(principal, rate, years)
    interest = principal * rate * years
    return interest
```

```python
assert calculate_interest(10_000, 0.05, 1) == 500
assert calculate_interest(2_000, 0.10, 3) == 600
print("✅ Bug 1 fixed")
```

## Bug 2 · Currency conversion

If **1 USD = 0.9 EUR**, then 100 USD should be **90 EUR**.
This one gives no error message. Run it and look at the answer.

```python
def convert_usd_to_eur(usd_amount, eur_per_usd):
    return usd_amount / eur_per_usd
```

```python
print("100 USD =", convert_usd_to_eur(100, 0.9), "EUR")
assert round(convert_usd_to_eur(100, 0.9), 2) == 90.0
print("✅ Bug 2 fixed")
```

## Bug 3 · Return on investment

ROI = (gain − cost) ÷ cost. Selling for 120 something that cost 100 is an ROI of **0.2** (20%).

```python
def calculate_roi(gain, cost):
    return (gains - cost) / cost
```

```python
assert round(calculate_roi(120, 100), 4) == 0.2
print("✅ Bug 3 fixed")
```

## Bug 4 · Is this trade profitable?

Should return `True` when revenue is greater than cost.

```python
def is_profitable(revenue, cost):
    return revenue <= cost
```

```python
assert is_profitable(5000, 3000) == True
assert is_profitable(3000, 5000) == False
assert is_profitable(4000, 4000) == False
print("✅ Bug 4 fixed")
```

## Bug 5 · Risk score

Risk score = volatility ÷ exposure. But what if exposure is **0**?
Decide what should happen. The test expects the function to return `None` (meaning "no score") when exposure is 0.

```python
def calculate_risk_score(volatility, exposure):
    return volatility / exposure
```

```python
assert calculate_risk_score(10, 5) == 2
assert calculate_risk_score(10, 0) is None
print("✅ Bug 5 fixed")
```

## 🏆 Boss level · Budget tracker (3 bugs)

This one has **three** bugs: one syntax, one runtime and one that makes the code **run forever**.

> **If a cell keeps running and never finishes**, press the **Interrupt** button (■ at the top of the notebook). That's how you stop an infinite loop.

```python
def get_total_spent(transactions):
    total = 0
    for amount in transactions:
        total += amount
    return total


def budget_status(spent, budget):
    if spent > budget
        return "Over budget"
    elif spent < budget:
        return "Within budget"
    else:
        return "Exactly on budget"


def days_until_over_budget(spent, budget, daily_spend):
    """How many more days of spending `daily_spend` before we go over budget?"""
    days = 0
    while spent <= budget:
        days += 1
    return days


transactions = [150, 200, "350", 100]
total = get_total_spent(transactions)
print("Spent so far:", total)
print("Status:", budget_status(total, 1000))
print("Days until over budget at 50/day:", days_until_over_budget(total, 1000, 50))
```

```python
assert get_total_spent(transactions) == 800
assert budget_status(800, 1000) == "Within budget"
assert budget_status(1200, 1000) == "Over budget"
assert days_until_over_budget(800, 1000, 50) == 5
print("🏆 Boss level cleared!")
```
